## Test : un signal sinusoïdal bruité

```python
t = np.linspace(0, 2, 1000)
signal = np.sin(2*np.pi*3*t) + 0.3*np.random.randn(len(t))
plt.plot(t, signal)
plt.xlabel('Temps (s)')
plt.ylabel('Amplitude')
plt.title('Sinusoïde bruitée')
plt.show()
```
